```python
import pandas as pd

ball = pd.read_csv('ball_by_ball_data.csv')
matches = pd.read_csv('ipl_matches_data.csv')
players = pd.read_csv('players-data-updated.csv')
teams = pd.read_csv('teams_data.csv')

# SRH team_id is 2
srh_matches = matches[(matches['team1'] == 2) | (matches['team2'] == 2)]
srh_ball = ball[(ball['team_batting'] == 2) | (ball['team_bowling'] == 2)]

print("SRH matches:", len(srh_matches))
print("SRH deliveries:", len(srh_ball))
print("Seasons:", sorted(srh_matches['season_id'].astype(str).unique()))
```

```python
srh_matches = srh_matches[srh_matches['season_id'] >= 2013]
srh_ball = srh_ball[srh_ball['season_id'] >= 2013]

print("SRH only matches:", len(srh_matches))
print("SRH only deliveries:", len(srh_ball))
print("Seasons:", sorted(srh_matches['season_id'].unique()))
```

```python
search_names = ['Warner', 'Dhawan', 'Williamson', 'Bhuvneshwar', 
                'Rashid', 'Abhishek', 'Klassen', 'Natarajan', 
                'Cutting', 'Steyn', 'Yuvraj', 'Pandey']

for name in search_names:
    found = srh_ball[srh_ball['batter'].str.contains(name, case=False, na=False)]['batter'].unique()
    if len(found) > 0:
        print(f"{name}: {found}")
    else:
        found = srh_ball[srh_ball['bowler'].str.contains(name, case=False, na=False)]['bowler'].unique()
        print(f"{name} (bowler): {found}")
```

```python
# SRH win/loss by season
srh_matches['srh_won'] = srh_matches['match_winner'] == 2
season_stats = srh_matches.groupby('season_id').agg(
    matches=('match_id', 'count'),
    wins=('srh_won', 'sum')
).reset_index()
season_stats['losses'] = season_stats['matches'] - season_stats['wins']
season_stats['win_pct'] = (season_stats['wins'] / season_stats['matches'] * 100).round(1)
print(season_stats)
```

```python
warner = srh_ball[srh_ball['batter'] == 'DA Warner']
print("Total runs:", warner['batter_runs'].sum())
print("Total balls:", len(warner[warner['is_wide_ball'] == False]))
print("Matches:", warner['match_id'].nunique())
print("Centuries:", warner.groupby('match_id')['batter_runs'].sum().apply(lambda x: x >= 100).sum())
print("Fifties:", warner.groupby('match_id')['batter_runs'].sum().apply(lambda x: 50 <= x < 100).sum())
```

```python
abhishek = srh_ball[srh_ball['batter'] == 'Abhishek Sharma']
print("Total runs:", abhishek['batter_runs'].sum())
print("Total balls:", len(abhishek[abhishek['is_wide_ball'] == False]))
print("Matches:", abhishek['match_id'].nunique())
print("Centuries:", abhishek.groupby('match_id')['batter_runs'].sum().apply(lambda x: x >= 100).sum())
print("Fifties:", abhishek.groupby('match_id')['batter_runs'].sum().apply(lambda x: 50 <= x < 100).sum())
sr = abhishek['batter_runs'].sum() / len(abhishek[abhishek['is_wide_ball'] == False]) * 100
print(f"Strike Rate: {sr:.1f}")
```

```python
mi_ball = ball[(ball['team_batting'] == 3)]  # MI team_id is 3

top_batters = mi_ball.groupby('batter')['batter_runs'].sum().sort_values(ascending=False).head(10)
print(top_batters)
```

```python
mi_ball = ball[ball['team_batting'] == 3]

mi_stats = mi_ball.groupby('batter').agg(
    runs=('batter_runs', 'sum'),
    balls=('batter_runs', 'count'),
    matches=('match_id', 'nunique')
).reset_index()

mi_stats['strike_rate'] = (mi_stats['runs'] / mi_stats['balls'] * 100).round(1)
mi_stats = mi_stats.sort_values('runs', ascending=False).head(10)
print(mi_stats.to_string())
```

```python
srh_matches_full = matches[
    ((matches['team1'] == 2) | (matches['team2'] == 2)) &
    (matches['season_id'] >= 2013)
].copy()

print(srh_matches_full['venue'].value_counts().head(10))
```

```python
venue_map = {
    'Rajiv Gandhi International Stadium, Uppal': 'Rajiv Gandhi Stadium',
    'Rajiv Gandhi International Stadium, Uppal, Hyderabad': 'Rajiv Gandhi Stadium',
    'Rajiv Gandhi International Stadium': 'Rajiv Gandhi Stadium',
    'Wankhede Stadium, Mumbai': 'Wankhede Stadium',
    'MA Chidambaram Stadium, Chepauk, Chennai': 'Chepauk Stadium',
    'MA Chidambaram Stadium, Chepauk': 'Chepauk Stadium',
    'Feroz Shah Kotla': 'Arun Jaitley Stadium',
    'Feroz Shah Kotla, Delhi': 'Arun Jaitley Stadium',
}

srh_matches_full['venue_clean'] = srh_matches_full['venue'].replace(venue_map)
srh_matches_full['venue_clean'] = srh_matches_full['venue_clean'].apply(
    lambda x: venue_map.get(x, x.split(',')[0].strip())
)

print(srh_matches_full['venue_clean'].value_counts().head(10))
```

```python
srh_matches_full['srh_won'] = srh_matches_full['match_winner'] == 2

venue_stats = srh_matches_full.groupby('venue_clean').agg(
    matches=('match_id', 'count'),
    wins=('srh_won', 'sum')
).reset_index()

venue_stats['win_pct'] = (venue_stats['wins'] / venue_stats['matches'] * 100).round(1)
venue_stats = venue_stats[venue_stats['matches'] >= 4]  # only venues with 4+ matches
venue_stats = venue_stats.sort_values('win_pct', ascending=False)

print(venue_stats.to_string())
```
